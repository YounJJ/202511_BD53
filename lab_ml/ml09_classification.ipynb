```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import SGDClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.datasets import load_breast_cancer
from sklearn.datasets import load_iris
```

# 2진 분류(binary-class classification) 문제

*   breast cancer 데이터에서 Malignant/Benign 분류
*   KNN classifier
*   Logistic regression
*   SGD classifier

```python
# 데이터 셋 로드
x, y = load_breast_cancer(return_X_y=True)
```

```python
# 데이터 셋 분리
x_train, x_test, y_train, y_test = train_test_split(x, y,
                                                    test_size=0.2, stratify=y)
```

```python
# KNN classifier
pipe = Pipeline(steps=[("scaler", StandardScaler()),
                       ("clf", KNeighborsClassifier())])
```

```python
pipe.fit(X=x_train, y=y_train)
```

```python
# KNN 훈련 셋 정확도
pipe.score(X=x_train, y=y_train)
```

```python
train_pred = pipe.predict(X=x_train)
train_pred[:6]
```

```python
confusion_matrix(y_true=y_train, y_pred=train_pred)
```

```python
# KNN 테스트 셋 정확도
pipe.score(X=x_test, y=y_test)
```

```python
# Logistic Regression
pipe = Pipeline(steps=[("scaler", StandardScaler()),
                       ("clf", LogisticRegression())])
```

```python
pipe.fit(X=x_train, y=y_train)
```

```python
# Logistic 훈련 셋 정확도
pipe.score(X=x_train, y=y_train)
```

```python
train_pred = pipe.predict(X=x_train)
train_pred[:6]
```

```python
confusion_matrix(y_true=y_train, y_pred=train_pred)
```

```python
# Logistic 테스트 셋 정확도
pipe.score(X=x_test, y=y_test)
```

```python
# SGD classifier
pipe = Pipeline(steps=[("scaler", StandardScaler()),
                       ("clf", SGDClassifier())])
```

```python
pipe.fit(X=x_train, y=y_train)
```

```python
# SGD 훈련 셋 정확도
pipe.score(X=x_train, y=y_train)
```

```python
train_pred = pipe.predict(X=x_train)
train_pred[:6]
```

```python
confusion_matrix(y_true=y_train, y_pred=train_pred)
```

```python
# SGD 테스트 셋 정확도
pipe.score(X=x_test, y=y_test)
```

# Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix, classification_report
```

## 데이터 셋 준비

```python
bc = datasets.load_breast_cancer()
```

```python
bc.keys()
```

```python
x = bc.data  # 특성 배열(2d array)
y = bc.target  # 타겟 배열(1d array)
```

```python
x.shape
```

```python
y.shape
```

```python
pd.Series(y).value_counts()  # 0 - Malignant(암), 1 - Benign(음성)
```

TODO: 탐색적 데이터 분석(데이터 시각화: 빈도 막대 그래프, 히스토그램, 산점도 pairplot)

훈련 셋, 테스트 셋 나누기

```python
x_train, x_test, y_train, y_test = train_test_split(x, y,
                                                    test_size=0.2,
                                                    random_state=42,
                                                    stratify=y)
```

## KNN

```python
pipe = Pipeline(steps=[("scaler", StandardScaler()),
                       ("knn", KNeighborsClassifier())])
```

```python
# ML 모델 훈련
pipe.fit(X=x_train, y=y_train)
```

```python
train_pred = pipe.predict(X=x_train)  # 훈련 셋 예측값
```

```python
cm_train = confusion_matrix(y_true=y_train, y_pred=train_pred)
cm_train
```

```python
sns.heatmap(data=cm_train, cmap="Blues", annot=True, fmt="d",
            xticklabels=bc.target_names, yticklabels=bc.target_names)
plt.show()
```

```python
print(classification_report(y_true=y_train, y_pred=train_pred, target_names=bc.target_names))
```

```python
test_pred = pipe.predict(X=x_test)  # 테스트 셋 예측값
```

```python
print(classification_report(y_true=y_test, y_pred=test_pred, target_names=bc.target_names))
```

## Logistic Regression

```python
# Logistic Regression ML 모델 생성
logit = LogisticRegression(random_state=42, max_iter=10_000)
```

```python
# ML 모델 훈련
logit.fit(X=x_train, y=y_train)
```

```python
logit.coef_
```

```python
train_pred = logit.predict(X=x_train)  # 훈련 셋 예측값
```

```python
cm_train = confusion_matrix(y_true=y_train, y_pred=train_pred)
cm_train
```

```python
print(classification_report(y_true=y_train, y_pred=train_pred,
                            target_names=bc.target_names))
```

```python
test_pred = logit.predict(X=x_test)  # 테스트 셋 예측값
```

```python
print(classification_report(y_true=y_test, y_pred=test_pred,
                            target_names=bc.target_names))
```

## SGD Classifier

```python
pipe = Pipeline(steps=[("poly", PolynomialFeatures(include_bias=False)),
                       ('scaler', StandardScaler()),
                       ('estimator', SGDClassifier(loss="log_loss", random_state=42))])
```

```python
# ML 모델 훈련
pipe.fit(X=x_train, y=y_train)
```

```python
pipe["estimator"].coef_  # 모델이 데이터를 학습한 후에 찾은 모델 파라미터들
```

```python
train_pred = pipe.predict(X=x_train)  # 훈련 셋 예측값
```

```python
cm_train = confusion_matrix(y_true=y_train, y_pred=train_pred)
cm_train
```

```python
print(classification_report(y_true=y_train, y_pred=train_pred,
                            target_names=bc.target_names))
```

```python
test_pred = pipe.predict(X=x_test)
print(classification_report(y_true=y_test, y_pred=test_pred,
                            target_names=bc.target_names))
```

# 다중 클래스 분류(multi-class classification) 문제

*   KNN
*   Logistic regression
*   SGD classifier

## 데이터셋 준비

```python
iris = datasets.load_iris()
```

```python
x = iris.data  # 특성 배열(2d array)
y = iris.target  # 타겟 배열, 레이블(1d array)
```

```python
iris.feature_names
```

```python
iris.target_names
```

```python
pd.Series(y).value_counts()
```

TODO: 탐색적 데이터 탐색(시각화 - countplot, histogram, pairplot, ...)

```python
# 훈련 셋, 테스트 셋 나누기
x_train, x_test, y_train, y_test = train_test_split(x, y,
                                                    test_size=0.2,
                                                    random_state=42,
                                                    stratify=y)
```

## KNN

```python
pipe = Pipeline(steps=[('scaler', StandardScaler()),
                       ('estimator', KNeighborsClassifier())])
```

```python
pipe.fit(X=x_train, y=y_train)  # ML 모델 훈련
```

```python
train_pred = pipe.predict(X=x_train)  # 훈련 셋 예측값
```

```python
cm_train = confusion_matrix(y_true=y_train, y_pred=train_pred)
cm_train
```

```python
sns.heatmap(data=cm_train, cmap="Blues", annot=True,
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.show()
```

```python
# 훈련 셋 분류 리포트
print(classification_report(y_true=y_train, y_pred=train_pred, target_names=iris.target_names))
```

```python
test_pred = pipe.predict(X=x_test)
```

```python
cm_test = confusion_matrix(y_true=y_test, y_pred=test_pred)
cm_test
```

```python
pipe.score(X=x_test, y=y_test)  # 테스트 셋 정확도
```

## Logistic Regression

```python
# ML 모델 생성
logit = LogisticRegression(random_state=42, penalty="l2", C=(1/0.0001), max_iter=1_000, tol=0.001)
```

```python
# ML 모델 훈련
logit.fit(X=x_train, y=y_train)
```

```python
# Logistic Regression 알고리즘이 찾은 모델 파라미터들
logit.coef_  # shape: (3, 4) = (n_classes, n_features)
```

```python
# 훈련 셋 예측값
train_pred = logit.predict(X=x_train)
```

```python
cm_train = confusion_matrix(y_true=y_train, y_pred=train_pred)
cm_train
```

```python
# 훈련 셋 정확도(accuracy)
logit.score(X=x_train, y=y_train)
```

```python
test_pred = logit.predict(X=x_test)
cm_test = confusion_matrix(y_true=y_test, y_pred=test_pred)
cm_test
```

```python
# 테스트 셋 정확도
logit.score(X=x_test, y=y_test)  #> 과소적합(under-fitting)
```

## SGD Classifier

```python
# SGD 분류 모델 생성
sgd = SGDClassifier(loss='log_loss', random_state=42)
```

```python
# 모델 훈련
sgd.fit(X=x_train, y=y_train)
```

```python
# SGD 알고리즘이 데이터를 학습하고 찾은 모델 파라미터
sgd.coef_
```

```python
# 훈련 셋 예측값
train_pred = sgd.predict(X=x_train)
```

```python
cm_train = confusion_matrix(y_true=y_train, y_pred=train_pred)
cm_train
```

```python
# 훈련 셋 정확도
sgd.score(X=x_train, y=y_train)
```

```python
# 테스트 셋 예측값
test_pred = sgd.predict(X=x_test)
```

```python
cm_test = confusion_matrix(y_true=y_test, y_pred=test_pred)
cm_test
```

```python
# 테스트 셋 정확도
sgd.score(X=x_test, y=y_test)
```

### Pipeline: PolynomialFeatures -> StandardScaler -> SGDClassifier

```python
pipe = Pipeline(steps=[('poly', PolynomialFeatures()),
                       ('scaler', StandardScaler()),
                       ('estimator', SGDClassifier(loss='log_loss', random_state=42))])
```

```python
pipe.fit(X=x_train, y=y_train)
```

```python
train_pred = pipe.predict(X=x_train)
cm_train = confusion_matrix(y_true=y_train, y_pred=train_pred)
cm_train
```

```python
# SGD가 데이터 학습에서 찾은 모델 파라미터들
pipe['estimator'].coef_
```

```python
pipe.score(X=x_train, y=y_train)  # 훈련 셋 정확도
```

```python
pipe.score(X=x_test, y=y_test)  # 테스트 셋 정확도
```
