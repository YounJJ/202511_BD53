# Quiz

    랜덤으로 메뉴선택(금액을 입력)
    저녁먹고 남은 돈을 저장
    차액에 따라 다음날 점심을 먹을 수 있다 없다 를 출력
    6,000 이상 식사 가능

```python
wallet = int(input("금액을 입력하세요 : "))

menu = [15_000, 8_000, 4_500]

import random
dinner = random.choice(menu)

if wallet >= dinner:
    wallet -= dinner
    print("저녁 식대는 {}원".format(dinner))
    if wallet >= 6_000:
        print("남은 돈이 {}원이므로 점심도 든든하게~".format(wallet))
    else:
        print("남은 돈이 {}원이므로 점심은 강제 다이어트..".format(wallet))
else:
    print("저녁 식대는 {}원".format(dinner))
    if wallet >= 6_000:
        print("남은 돈이 {}원이므로 저녁은 못먹지만 점심은 든든하게~".format(wallet))
    else:
        print("남은 돈이 {}원이므로 점심도 강제 다이어트..".format(wallet))
```
