# 조건문(condition)
    특정 조건에 따라 코드를 실행하고자 할 때 사용
    if, else, elif

## if

    if 조건문 :
        조건이 성립할 때 실행될 구문
        
    조건문에 boolean 타입 외의 데이터 타입이 오면 boolean 타입으로 형변환되어 판단

```python
if True: # if 조건문:
    print("It's True")
print("Done!")
```

```python
num = 0
if num:
    print("It's False")
print("Done!")
```

    int : 0을 제외한 나머지 숫자는 모두 True
    float : 0.0을 제외한 나머지 숫자는 모두 True
    str : ""을 제외한 나머지는 모두 True
    list, tuple, dict : [], (), {}을 제외한 나머지는 모두 True

## if else

    if 조건문:
        조건문이 성립할 때 실행되는 구문
    else:
        조건문 외의 경우가 성립할 때 실행되는 구문

```python
if False:
    print("It's false")
else:
    print("True")
```

```python
num = int(input("숫자를 입력하세요"))
```

```python
num = int(input("숫자를 입력하세요 : "))
if num % 2:
    print("홀수입니다")
else:
    print("짝수입니다")
```

## if elif else
    if 조건문1:
        실행문1
    elif 조건문2: #몇 개든 만들 수 있음!
        실행문2
    else:
        실행문3

    if 조건문부터 순차적으로 확인한 후 True의 경우일 때 해당 조건의 실행문이 실행된다.

```python
# 잔고에 따라 저녁메뉴 선택하기
wallet = int(input("금액을 입력하세요 : "))

if wallet >= 20000:
    print("오늘 저녁은 치킨이닭!")
elif wallet >= 10000:
    print("오늘 저녁은 치킨덮밥이닭!")
elif wallet >= 5000:
    print("오늘 저녁은 치킨버거닭!")
elif wallet >= 2000:
    print("오늘 저녁은 치킨마요 김밥이닭..")
else:
    print("굶어야지..")
```

## 상황연산자

    한줄의 코드로 if, else를 표현할 수 있는 방법

    실행문1 if 조건문1 else 실행문2

```python
data = 1
if data:
    print("not empty")
else:
    print("zero")
```

```python
data = 0
print("not zero") if data else print("zero")
```
