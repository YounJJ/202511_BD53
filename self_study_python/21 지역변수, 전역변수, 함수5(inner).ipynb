## 지역변수와 전역변수

*   함수 안에서 선언되는 변수와 밖에서 선언되는 변수의 범위
*   global : 전역
*   local : 지역

```python
global_v = 100

def exam():
    print(global_v)
```

```python
exam()
```

```python
global_v = 100

def exam2():
    global_v = 20
    print(global_v)
```

```python
exam2()
```

```python
global_v
```

```python
def exam3():
    global global_v
    global_v = 50
    print(global_v)
```

```python
exam3()
```

```python
global_v
```

## 함수안에 함수가 선언되는 경우(inner function)

*   함수를 지역영역에 선언 -> 함수 안에 함수

```python
def outer(a, b):
    def inner(c, d):
        return c + d
    return inner(a, b)
```

```python
outer(1, 2)
```

```python
inner(1, 2)
```

```python
def outer2(a, b):
    def inner(c, d):
        return c + d
    return inner
```

```python
outer2(1, 2)
```

```python
outer2(3, 4)(1, 2)  # outer2(3, 4) (1, 2) -> inner(1, 2)
```

```python
outer2(100, 1000)(3, 4)
```
