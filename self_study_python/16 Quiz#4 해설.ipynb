### Quiz
    이중 for문을 이용하여 2차원 배열의 원소들 합을 1차원으로 나타내기

$$
    \begin{pmatrix}
    2 & 4 \\
    6 & 8 \\
    \end{pmatrix}
    +
    \begin{pmatrix}
    1 & 3 \\
    5 & 7 \\
    \end{pmatrix}
    =
    \begin{pmatrix}
    3 & 7 \\
    11 & 15 \\
    \end{pmatrix}
$$

```python
m1 = [[2, 4], [6, 8]]
m2 = [[1, 3], [5, 7]]

result = []

for i in range(len(m1)):  # i = 0, 1
    for j in range(len(m2)):  # j = 0, 1
        result.append(m1[i][j] + m2[i][j])





print(result)
```

```python
result = [[0, 0], [0, 0]]
m1 = [[2, 4], [6, 8]]
m2 = [[1, 3], [5, 7]]

for i in range(len(m1)): # i = 0, 1
    for j in range(len(m2)): # j = 0, 1
        result[i][j] = m1[i][j] + m2[i][j]

print(result)
```
