#### dict형 데이터 타입 key, value 값 확인
    keys() : key값 확인
    values() : value값 확인
    items() : key, value 쌍으로 확인
    * 주의
        순서가 없어서 index 사용 불가

```python
dic.keys(), dic.values(), dic.items()
```

```python
for key in dic.keys():
    print(key)
```
