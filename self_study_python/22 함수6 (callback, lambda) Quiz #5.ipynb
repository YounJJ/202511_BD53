## callback function

*   함수를 args 파라미터로 설정해서 사용
*   함수가 다른 함수를 호출하여 결과값을 실행
    *   재귀함수에서 잘 사용

```python
def calc(func, a, b):
    return func(a, b)
```

```python
def hap(a, b):
    return a + b
def gop(a, b):
    return a * b
```

```python
calc(hap, 3, 4)
```

```python
calc(gop, 3, 4)
```

## lambda function

*   파라미터를 간단한 계산으로 리턴

```python
def hap(a, b):
    return a + b
```

```python
hap2 = lambda a, b: a + b  # 파라미터 : 리턴
hap2(4, 5)
```

## string 타입 데이터를 입력 받았을 때 문자열 가운데를 출력

```python
# easyupclass -> p

text = input("type text : ")
disc = int(len(text)/2)
print("text : ", text)
print("median : " , disc)
```

```python
substr = lambda x : x[disc] if len(x)%2 else x[disc-1:disc+1]
substr
```

```python
substr(text)
```
