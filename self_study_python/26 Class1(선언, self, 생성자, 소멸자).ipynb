## Basic class

### class 선언

```python
class Account:

    loan = 0
    deposit = 0

    def save(self, money):
        self.deposit += money
        print("현재 계좌의 잔액은 {}원 입니다.".format(deposit))

    def withdraw(self, money):
        self.deposit -= money
        print("인출 금액은 {}원 입니다. 현재 계좌의 잔액은 {}원 입니다.".format(money, deposit))

    def debt(self, d_money):
        self.deposit += d_money
        self.credit_debt = d_money + self.loan
        return -int(self.credit_debt)
```

```python
acc = Account()
```

```python
acc.loan
```

### 생성자
*   클래스의 인스턴스 객체가 생성될 때 자동으로 호출되는 메소드

```python
class Account:

    def __init__(self):
        self.deposit = 0
        self.loan = 0
        print("계좌가 개설되었습니다!")

    def save(self, money):
        self.deposit += money
        print("현재 계좌의 잔액은 {}원 입니다.".format(deposit))

    def withdraw(self, money):
        self.deposit -= money
        print("인출 금액은 {}원 입니다. 현재 계좌의 잔액은 {}원 입니다.".format(money, deposit))

    def debt(self, d_money):
        self.deposit += d_money
        self.credit_debt = d_money + self.loan
        return -int(self.credit_debt)
```

```python
acc3 = Account()
```

```python
acc3.loan
```

```python
acc3.deposit
```

```python
acc3.loan = 500_000
acc3.loan
```

```python
acc3.debt(500_000)
```

### 소멸자
*   객체가 사라질 때 자동으로 호출되는 함수

```python
class Account:

    def __init__(self):
        self.deposit = 0
        self.loan = 0
        print("계좌가 개설되었습니다!")

    def save(self, money):
        self.deposit += money
        print("현재 계좌의 잔액은 {}원 입니다.".format(deposit))

    def withdraw(self, money):
        self.deposit -= money
        print("인출 금액은 {}원 입니다. 현재 계좌의 잔액은 {}원 입니다.".format(money, deposit))

    def debt(self, d_money):
        self.deposit += d_money
        self.credit_debt = d_money + self.loan
        return -int(self.credit_debt)

    def __del__(self):
        print("계좌가 폐쇄되었습니다. 이용해주셔서 감사합니다.")
```

```python
acc4 = Account()
```
