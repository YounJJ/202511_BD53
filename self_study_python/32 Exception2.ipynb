## try except finally

*   except 발생 유무와 상관없이 finally 구문은 실행

```python
print("Start")
def dividing():
    n1, n2 = tuple(input("insert numbers: ").split(" "))
    try:
        return int(n1)/int(n2)
    except:
        print("연산을 수행할 수 없습니다.")
    finally:
        print("연산 종료")
```

```python
print("Start")
dividing()
```

```python
print("Start")
dividing()
```

## try except Exception

*   python에서 발생 가능한 예외상황을 exception 객체에 미리 정의되어 있음
*   해당 Exception 발생 시 return에 원하는 사항을 작성할 수 있다

```python
def looping():
    n = input()
    try:
        for i in range(int(n) * 100_000):
            print(i)
    except KeyboardInterrupt:
        return "요청에 따라 종료"
```

```python
print("Start")
looping()
print("Done!")
```

```python
def dividing():
    n1, n2 = tuple(input("insert numbers: ").split(" "))
    try:
        return int(n1)/int(n2)
    except ZeroDivisionError:
        print("나눌 수 없습니다!")
```

```python
print("Start")
dividing()
print("Done!")
```

## 사용자 정의 예외처리

*   python은 정상으로 인식할 수 있으나 코드를 작성하는 작성자 기준에서 오류라고 판단되는 부분을 예외처리

```python
class LowNumber(Exception):
    def __str__(self):
        return "Number is greater than 10"
```

```python
def input_number(num):
    if num <= 10 : raise LowNumber()
    print(num)
```

```python
input_number(77)
```

```python
input_number(9)
```
