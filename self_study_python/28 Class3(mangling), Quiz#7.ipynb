## mangling

*   클래스 변수에 접근할 수 없도록 설정하는 방법
*   변수명 앞에 __(언더바 2개)를 붙여서 사용

```python
class Employee:
    mileage = 0
    rate = round(1/15, 2)

    def __init__(self, first, last, salary):
        print("입사를 축하합니다!")
        self.first = first
        self.last = last
        self.salary = salary
        self.id = first.lower() + last.lower() + "@euc.com"
        self.mileage = 100_000
        self.__rate = round(1/15, 2)

    def save_mileage(self, point):
        self.mileage += int(point + (self.salary) * self.__rate)
        print("추가 포인트는 {}점, 누적 마일리지는 {}점 입니다.".format(point, self.mileage))
```

```python
emp4 = Employee('bekko', 'A', 1_000)
```

```python
emp4.rate
```

```python
emp4.rate = 0.7
emp4.rate
```

```python
emp4.save_mileage(10_000)
```

```python
emp4.__rate = 0.7
emp4.__rate
```

```python
emp4.save_mileage(10_000)
```

# Quiz

*   사용처에 따라 적립률이 다른 카드 종류 만들기
*   1개의 슈퍼클래스, 여러개의 서브 클래스 생성
*   특정 서브클래스는
    *   영화관, 극장, moive등 특정 키워드가 들어갔을 경우 적립률 변경
    *   그 외의 장소 혹은 키워드가 없을 경우 기존 적립률 유지
    *   mangling 방식을 이용하여 적립률을 변경 못하게 만들기!

```python
class Card:
    point = 0
    rate = round(1/15, 2)

    def __init__(self):
        print("가입을 축하합니다!")
        self.__rate = round(1/15, 2)

    def save_point(self, spend):
        self.point += int(spend * self.__rate)
        print(f"추가 포인트는{int(spend * self.__rate)}점, 총 포인트는 {self.point}점 입니다.")
```

```python
class Mcard(Card):

    def __init__(self):
        venue = input("사용처를 입력해주세요 : ")
        if "영화" in venue or "극장" in venue or "movie" in venue:
            self.__rate = round(1/8, 2)
        else:
            self.__rate = round(1/15, 2)
        print("가입을 축하합니다!")

    def save_point(self, spend):
        self.point += int(spend * self.__rate)
        print(f"추가 포인트는{int(spend * self.__rate)}점, 총 포인트는 {self.point}점 입니다.")
```

```python
kb = Mcard()
```

```python
kb.save_point(50_000)
```

```python
mk = Card()
```

```python
mk.save_point(50_000)
```
