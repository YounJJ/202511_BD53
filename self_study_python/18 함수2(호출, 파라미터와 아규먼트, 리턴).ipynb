### 호출

#### parameter, argument
    def 함수명(파라미터):
        return

    함수명(아규먼트)

*   파라미터는 함수를 선언할 때 호출할 편수
*   아규먼트는 함수를 호출할 때 보내줄 변수
*   함수를 선언할 때 파라미터의 타입과 개수를 맞추지 않으면 에러가 발생

```python
def bowling(n1, n2, adv=50):
    return n1+n2+adv

bowling(200, 150)
```

```python
bowling(105, 120, 0)
```

### return

*   선언한 함수의 결과값을 저장
*   함수내에서 연산 진행중에 return에 도달하면 연산이 종류
*   return을 작성하지 않으면 결과값을 저장하지 않음

```python
def print_test(text):
    print(text.upper())
    return text.upper()

text = "Hello Python"
print_test(text)
```

```python
def print_test(text):
    print(text.upper())
#    return text.upper()
print_test(text).lower()
```

```python
def print_test2(text):
    print(text.upper())
    return text.upper()

text = "Hello Python"
print_test2(text).lower()
```

```python
# 함수 연산을 중단시키는 예제
def alpha_check(text):
    for t in text:
        if t.isalpha(): # isalpha(): str타입의 변수가 알파벳으로 구성되어있으면 True, 아니면 False를 반환
            continue
        else:
            print(text.index(t)+1)  # index(): index값 반환
            return False

    print("Clear!")
    return True
```

```python
text1 = "abcdefgh1jkl"
text2 = "eazyupc1ass"
text3 = "Adrian"

alpha_check(text1)
```

```python
alpha_check(text2)
```

```python
alpha_check(text3)
```
