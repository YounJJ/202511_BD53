# super()

*   부모(슈퍼) 클래스에서 사용된 함수의 코드를 받아서 자식(서브) 클래스에서 재사용 시 사용

```python
class Phone:
    def __init__(self, function = None):
        self.name = function
        print("{} phone 개통.".format(self.name))

    def calling(self):
        print("phone dial selected")
```

```python
class CameraPhone(Phone):
    def __init__(self, function = None):
        super().__init__(function)

    def camera(self):
        print("camera selected")
```

```python
ph = Phone()
```

```python
ph.calling()
```

```python
cam = CameraPhone("Camera")
```

```python
cam.calling()
```

```python
cam.camera()
```

```python
class WebPhone(CameraPhone):
    def __init__(self, function = None):
        super().__init__(function)

    def web(self):
        print("Web browser open")
```

```python
web = WebPhone("Web")
```

```python
web.calling()
```

```python
web.camera()
```

```python
web.web()
```

# getter, setter
*   객체의 내부 변수에 접근할 때 특정 로직을 거쳐서 접근시키는 방법
*   제약조건을 만들어서 접근 가능 여부를 만들 때 사용

```python
class User:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def setter(self, first_name):
        if len(first_name) >= 2:
            self.first_name = first_name
        else:
            print("error")

    def getter(self):
        return self.first_name[0].upper() + self.first_name[1:].lower() + " " + self.last_name[0].upper() + self.last_name[1:].lower()

    name = property(getter, setter)
```

```python
user1 = User('euc', 'uni')
```

```python
user1.name
```

```python
user1.name = "bekko"
```

```python
user1.name
```

```python
user1.name = "A"
user1.name
```

# 4 is a / has a
*   is a : 상속을 이용해서 클래스를 생성
*   has a : 객체를 가지고 클래스를 생성

```python
# is a
class User:
    def __init__(self, name, email):
        self.name = name
        self.email = email

class Id(User):
    def info(self):
        print(self.name, self.email)
```

```python
user = Id("bekko", "bekko@euc.com")
```

```python
user.info()
```
