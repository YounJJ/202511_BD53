```python
score = 0

test = [100, 90, 50, 65, 70]

for s in test:
    score += s

print(score)
```

## 함수의 선언과 호출

### 선언
    def 함수명(param1, param2, ...):
        # 작성하고 싶은 코드 구문
        #..
        #..
        return 결과값

```python
def score(test, point=0):
    for s in test:
        point += s

    return point
```

```python
test1 = [100, 50, 75, 65, 30]
test2 = [50, 43, 25, 15, 11, 20]

score(test1)
```

```python
score(test2, 320)
```

평균 점수에 따라 등급을 결정하는 함수 작성

```python
def grade(subj):
    score = 0
    gpa = 0

    for s in subj:
        score += s
    gpa = score/ len(subj)

    if gpa >= 4.3:
        return round(gpa, 2), "A+"
    elif gpa >= 4.0:
        return round(gpa, 2), "A"
    elif gpa >= 3.3:
        return round(gpa, 2), "B+"
    elif gpa >= 3.0:
        return round(gpa, 2), "B"
    elif gpa >= 2.3:
        return round(gpa, 2), "C+"
    elif gpa >= 2.0:
        return round(gpa, 2), "C"
    else:
        return round(gpa, 2), "F"
```

```python
mid = [3.3, 4.0, 3.3, 4.3, 2.3, 3.0, 3.0]
fin = [1.3, 0, 2.3, 2.0, 2.0, 2.3, 0, 1.0]

grade(mid)
```

```python
grade(fin)
```
