## Series

```python
import numpy as np
import pandas as pd
```

### Series 선언

```python
test = pd.Series(np.random.randint(10, size = 5))
test
```

```python
test2 = pd.Series({"name": "easyup",
                   "class": "python",
                   "category": "IT"})
test2
```

```python
test2[0]
```

```python
test3 = pd.Series({"name": "easyup",
                   "class": "python",
                   "age": 20})
test3
```

```python
type(test3[2])
```

```python
type(test3[0])
```

### index(행) 이름 설정

```python
test4 = pd.Series(np.random.randint(10, size = 5), index = range(1, 6))
test4
```

```python
test4[1]
```

```python
test5 = pd.Series(np.random.randint(10, size = 5), index = list('ABCDE'))
test5
```

```python
test5.index, test5.values
```

```python
scalar_test = pd.Series(10, index=range(1, 10))
scalar_test
```
