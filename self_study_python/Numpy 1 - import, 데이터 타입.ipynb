## 데이터 수정

```python
ls = [1, 2, 3]
ls[1] = 5
ls
```

```python
ary3 = np.array([[1, 2, 3], [0, 0, 0]])
ary3[1, 1] = 1
ary3
```

```python
ary3[1] = [1, 2, 3]
ary3
```

## array 브로드캐스팅
*   차원이 다른 연산을 동일한 차원으로 자동으로 변경시켜 연산을 가능하게 해주는 것

```python
ary3[0] = 0
ary3
```

```python
ls * 10
```

```python
ary3 * 10
```

## array 조건 선택

```python
ary3
```

```python
ary3[0] = 5
ary3[1] = [3, 4, 5]
ary3
```

```python
# mask 생성
idx = ary3 < 5
idx
```

```python
ary3[idx]
```

## 조건에 맞는 값만 변경

```python
ary3[idx] = 100
ary3
```
