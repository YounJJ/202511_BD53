*   사용처에 따라 적립률이 다른 카드 종류 만들기
*   1개의 슈퍼클래스, 여러개의 서브 클래스 생성
*   특정 서브클래스는
    *   영화관, 극장, moive등 특정 키워드가 들어갔을 경우 적립률 변경
    *   그 외의 장소 혹은 키워드가 없을 경우 기존 적립률 유지
    *   mangling 방식을 이용하여 적립률을 변경 못하게 만들기!

```python
class Card:  # super class
    def __init__(self, lim = 500_000):
        self.deposit = 0
        self.point = 0
        self.debt = 0
        self.limit = lim
        print("카드가 발급되었습니다. 초기한도 {}원".format(self.limit))

    def use_card(self, cost):
        if cost <= self.limit:
            self.debt += cost
            self.point += int(cost*0.02)
            print("{}원 사용. 결제 예정금액 {}원.".format(cost, self.debt))
            print("{}점 적립되었습니다. 잔여 포인트 {}점.".format(int(cost*0.02), self.point))
        else:
            print("결제 한도 초과! 잔여 한도 {}원.".format(self.limit - self.debt))

        def payment(self, money = 0):
            self.deposit += money
            if self.debt > self.deposit:
                print("{}원 결제. 남은 결제금액 {}원.".format(self.deposit, self.debt - self.deposit))
                self.debt =- self.deposit
                self.deposit = 0
            else:
                print("{}원 결제.".format(self.debt))
                self.deposit -= self.debt
                self.debt = 0
```
