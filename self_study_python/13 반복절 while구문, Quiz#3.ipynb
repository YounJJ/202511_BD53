#### while 문

    while 조건문: # 조건문이 False가 될 때까지 반복

```python
rep = 10
while rep:
    print(rep)
    rep -= 1
```

#### while문 quiz
    list, dict, while문을 이용하여 코드를 만들어보기

```python
# 세뱃돈 구하기
# 총 세뱃돈의 금액 구하고, 평균 세뱃돈 금액 구하기
# 할아버지, 할머니, 아빠, 엄마로 구성된 리스트, 10만, 5만, 3만, 1만 -> dict형태로 구성
# len(), zip()
# while 문 작성
```

```python
# 리스트 작성
family = ["할아버지", "할머니", "아빠", "엄마"]
money = [100_000, 50_000, 30_000, 10_000]

# dict 만들기
bow_dict = dict(zip(family, money))

# while문 작성
cnt = len(family) -1
bow_money = 0

while cnt != -1:
    print("{}께서 {}원을 주셨다!".format(family[cnt], money[cnt]))
    bow_money += money[cnt]
    cnt -= 1

print("올해 세뱃돈은 {}원이다! 올해 평균 세뱃돈은 {}원이야!".format(bow_money, int(bow_money/len(family))))
```

#### while문의 무한루프

```python
res = 1
while res:
    print(res)
    res += 1
    # 탈출문 작성
    if res > 10:
        print("break!")
        break
```

### Quiz
    로또 번호 생성기

    1~45까지의 숫자를 중복없이 6개를 추출해서 한 리스트에 담기
    random 모듈을 이용해서 작성
    무한루프로 돌리되 break를 이용하여 탈출할 수 있게 작성
    멤버 연산 in, not in

```python
import random
i = random.randint(1, 45)
ls = [i]

while ls:
    i = random.randint(1, 45)
    if len(ls) == 6:
        break
    elif i not in ls:
        ls.append(i)

ls
```
