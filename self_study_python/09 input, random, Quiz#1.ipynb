## format 함수
    변수의 데이터를 문자형으로 변환하여 출력할 수 있는 방법

```python
account = 100_000
draw = 30_000
account -= draw

print("인출 금액은 {}원 입니다. 현재 잔액은 {}원 입니다.".format(draw, account))
```

# Quiz

    랜덤으로 메뉴선택(금액을 입력) #리스트를 이용하여 작성 random모듈
    저녁먹고 남은 돈을 저장 #남은 돈은 다른 변수에 저장
    차액에 따라 다음날 점심을 먹을 수 있다 없다 를 출력 # 조건문
    6,000 이상 식사 가능

```python
wallet = int(input("현재 지갑의 잔액을 10,000원 이상으로 입력하세요 : "))
dinner = [2_000, 4_000, 6_000, 8_000, 10_000]

import random
payment = random.choice(dinner)
wallet -= payment
print("오늘 저녁 식비는 {}원 입니다. 남은 잔액은 {}원 입니다.".format(payment, wallet))

if wallet >= 6_000:
    print("내일 점심 먹을수 있다!")
else:
    print("내일 점심 굶어야 한다!")
```
