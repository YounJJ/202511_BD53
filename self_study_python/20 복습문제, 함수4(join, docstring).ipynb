# Quiz

text 문장의 대소문자를 맞춰주는 함수를 생성하여 출력
예:

    python IS very FUN and EAsy
    -> Python is very fun and easy.
    문장의 마지막에 "."은 있을 수도 없을 수도..

```python
def typing(args):  # 이건 내가 만든거
    args = args.lower()
    args[0].upper() + args[1:]
    return args[0].upper() + args[1:] + "."

typing('python IS very FUN and EAsy')
```

```python
def correction(text):  # 이건 강사님이 만든거
    sentence = ""
    for i in range(len(text)):
        if not i:
            w = text[i].upper()
        else:
            w = text[i].lower()

        sentence += w
        # bprint(sentence)

    if sentence [-1] != ".":
        sentence += "."
    return sentence
```

```python
text = str(input("input sentence : "))
correction(text)
```

## join 문법

*   하나의 문자열로 합칠 수 있는 함수

```python
ls = ["python", "is", "very", "fun"]
" ".join(ls)
```

```python
"-".join(ls)
```

```python
def correction2(text):
    sentence = ''.join(text[i].upper() if not i else text[i].lower() for i in range(len(text)))
    if sentence[-1] != ".": sentence += "."
    return sentence
```

```python
text = str(input('input sentence : '))
correction2(text)
```
