## apply

*   map과 유사한 함수

### domain column 생성해보기

```python
def str_domain(email):
    return email.split("@")[1].split(".")[0]
```

```python
str_domain("adrian@gmail.co.kr")
```

```python
str_domain(df.loc[1].email)
```

```python
df['domain'] = df['email'].apply(str_domain)
df
```

## 수업용 데이터 생성 함수

```python
import random
```

```python
def set_names():
    names = [
        'akali', 'amumu', 'ashe', 'brand', 'braum',
        'caitlyn', 'darius', 'ekko', 'fiora', 'fiddlesticks',
        'galio', 'garen', 'graves', 'jax', 'jhin',
        'karma', 'kindred', 'leona', 'lulu', 'lux'
    ]
    return random.choice(names)

def set_level(start = 1, end = 18):
    return np.random.randint(start, end + 1)

def datas(n = 20):
    lists = []
    for _ in range(n):
        dic = {
            "Level": set_level(), "Name": set_names()
        }
        lists.append(dic)
    return lists
```

```python
df1 = pd.DataFrame(datas(5))
df2 = pd.DataFrame(datas(5))

df1, df2
```

### DataFrame 합치기

concat

*   row, columm으로 dataframe을 합칠 때 사용

```python
df3 = pd.concat([df1, df2])
df3
```

### index 재정렬

```python
df3.reset_index(drop = True, inplace = True)
df3
```

```python
df4 = pd.concat([df1, df2]).reset_index(drop = True)
df4
```

```python
pd.concat([df1, df2], axis = 1)
```

### group by

*   특정 컬럼의 중복되는 데이터를 합쳐서 새로운 dataframe을 생성

```python
df5 = pd.DataFrame(datas())
df5
```

### size

```python
res = df5.groupby("Name").size().reset_index(name = "counts")
res
```

### sort_values: 설정값으로 column 정렬

```python
res2 = res.sort_values(['counts'], ascending = False).reset_index(drop = True)
res2
```

### agg
*   파라미터로 'max', 'min', 'mean' 등을 사용

```python
df5.groupby("Name").agg("min").reset_index()
```

```python
df5.sort_values("Name")
```
