### Quiz
    로또 번호 생성기

    1~45까지 숫자를 중복없이 6개를 추출해서 한 리스트에 담기
    random 모듈을 이용해서 작성
    무한루프로 돌리되 break를 이용하여 탈출할 수 있게 작성
    멤버 연산 in, not in

```python
import random

lotto = []

for _ in range(5):
    one = []
    while True:
        number = random.randint(1, 45)
        if number not in one:
            one.append(number)
        one.sort()
        if len(one) == 6:
            break
    lotto.append(one)

lotto
```
