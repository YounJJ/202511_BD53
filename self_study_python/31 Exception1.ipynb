# 예외처리
*   프로그램이 실행 중 오류가 발생했을 때 -> 예외상황
*   예외상황이 발생하면 코드 진행이 중단되므로 계속 진행할 수 있도록 선언

## try exception

```python
try:
    문제가 없을 시 실행할 코드
except:
    문제 발생(예외상황)시 실행할 코드
```

````python
# 강제로 문제 발생 예제
def dividing2():
    n1, n2 = tuple(input("insert numbers : ").split(" "))
    return int(n1)/int(n2)
````

````python
print("Start")
dividing2()
print("Done")
````

````python
def dividing():
    n1, n2 = tuple(input("insert numbers : ").split(" "))
    try:
        return int(n1)/int(n2)
    except:
        print("나눗셈을 시행할 수 없습니다.")
````

````python
print("Start")
dividing()
print("Done")
````

## try except else

*   except가 실행되지 않으면 else의 실행문이 마지막에 출력

````python
print("Start")
n = input("set range: ")
try:
    res = sum(range(int(n)+1))
except:
    print("잘못된 입력입니다.")
else:
    print("result: ", res)

print("Done!")
````
