#### zip을 이용한 받아주기

```python
corp = ['google', 'facebook', 'apple', 'tesla']
prod = ['chrome', 'libra', 'iphone', 'e-car']
rank = [1, 3, 2, 4]

for c, p, r in zip(corp, prod, rank):
    print(c, p, r)
    # print(p)
```

#### for문 continue
    - continue를 사용 : for문의 조건문으로 돌아가서 루프를 실행(다음 순서로 진행)

```python
ls = [1, 2, 3, 4, 5]
for i in ls:
    if i%2:
        continue
        print("홀수는 슈퍼패스")
    print(i)
```

#### 중첩 for문

```python
# 구구단 전체 출력해보기
for i in range(1, 10):
    for j in range(2, 10):
        print("{} x {} = {}".format(j, i, j*i), end="\t")
    print()
```

#### list comprehension
    - list 데이터를 만들 때 for문을 이용하여 생성
    - 속도가 빠르고 ram을 적게 사용


```python
ls = [11, 12, 13, 14]
ls2 = ls ** 2
```

```python
ls2 = []
for i in ls:
    ls2.append(i ** 2)
ls2
```

```python
ls3 = [i ** 2 for i in ls]
ls3
```

### Quiz
    이중 for문을 이용하여 2차원 배열의 원소들 합을 1차원으로 나타내기

$$
    \begin{pmatrix}
    2 & 4 \\
    6 & 8 \\
    \end{pmatrix}
    +
    \begin{pmatrix}
    1 & 3 \\
    5 & 7 \\
    \end{pmatrix}
    =
    \begin{pmatrix}
    3 & 7 \\
    11 & 15 \\
    \end{pmatrix}
$$

```python
m1 = [[2, 4], [6, 8]]
m2 = [[1, 3], [5, 7]]

result = []

for i in range(0, 2):
    for j in range(0, 2):
        result.append(m1[i][j] + m2[i][j])





print(result)  # should be [[3, 7], [11, 15]]
```
