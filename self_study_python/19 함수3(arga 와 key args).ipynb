### arguments 와 keyword arguments

*   함수 선언시 파라미터의 수를 특정할 수 없을 경우 사용
*   keyword args의 경우 아규먼트로 함수 호출 시 키워드를 작성하여 호출
*   args는 튜플 타입
*   kwargs는 딕셔너리 타입

```python
def hap(*args, **kwargs):
    print(type(args), args)
    print(type(kwargs), kwargs)
    return sum(args) + sum(list(kwargs.values()))

hap(1, 2, 3, 4, num1=5, num2=6)
```

#### list타입 데이터를 args로 호출하는 방법

```python
def gop(n1, n2, n3):
    return n1 * n2 * n3

ls = [2, 3, 4]
gop(*ls)
```

```python
ls2 = [2, 3, 4, 5]
gop(*ls2)
```

```python
def hap2(*args):
    return sum(args)

ls3 = [3, 4, 5, 6]
hap2(*ls3)
```

```python
hap2(ls3)
```
