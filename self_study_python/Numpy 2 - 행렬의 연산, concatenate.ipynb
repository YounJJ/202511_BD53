# numpy로 작성하는 행렬



기본적인 list 연산도 구현 가능?

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[4, 5, 6], [1, 2, 3]]
A + B  #?
```

list에서 요소들의 합 구현은 쉽지 않음

```python
res = [[c + d for c, d in zip(a, b)] for a, b in zip (A, B)]
res
```

```python
import numpy as np
a = np.array(A)
b = np.array(B)
a + b
```

# 행렬의 곱

굳이 list로 만들어서 하겠다면...

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[6, 5], [4, 3], [2, 1]]
res = [[sum(a*b for a, b in zip (A_row, B_col)) for B_col in zip(*B)]for A_row in A]
res
```

zip(*)

```python
ls = [[1, 2, 3], [4, 5, 6]]
ls
```

```python
# Transpose
ls2 = list(map(list, zip(*ls)))
ls2
```

## numpy 곱

```python
import numpy as np
a = np.array(A)
b = np.array(B)

np.dot(a, b)  # = a@b
```

# Quiz
*   위 행렬의 곱을 for문으로 작성하면?

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[6, 5], [4, 3], [2, 1]]
res = [[sum(a*b for a, b in zip (A_row, B_col)) for B_col in zip(*B)]for A_row in A]
res
```

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[6, 5], [4, 3], [2, 1]]

sum_array= []
res = []
cal = 0

for A_row in A:
    sum_array = []
    for B_col in zip(*B):
        for a,b in zip(A_row, B_col):
            cal += a * b
        sum_array.append(cal)
        cal = 0

    res.append(sum_array)


print(res)
```
