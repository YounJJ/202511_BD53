### Loop문

    - 특정 코드를 종료시점까지 반복적으로 수행할 수 있는 구문
    - for, while, continue, break

#### for문 1

    - iterable한(반복 가능한) 값을 꺼내와서 변수에 대입시킨 후 갯수만큼 실행

        for 변수 in 범위:
            실행문

    - 범위(range)를 정하여 반복 횟수를 지정하고 지정한 횟수가 되면 loop문을 종료

```python
for i in range(10): # 출력하는 값은 9까지 나올 것임
    print(i)
```

```python
# 구구단 7단 출력해보기
for i in range(1, 10):
    print("7 x", i, "=", 7*i)
```

```python
text = "Hello python. Give me your world!"
for i in text:
    print(i)
```

```python
dic = {
    "hi":"hello",
    "Everyone":"world",
    "nice":"python"
}
for i in dic:
    print(i)
```

```python
for value in dic.values():
    print(value)
```

#### for문 quiz

```python
# 최대공약수 구하기
a = int(input("첫번째 숫자 : "))
b = int(input("두번째 숫자 : "))

for i in range(max(a, b), 0, -1): #range 함수에서 -1은 역순으로 돌아간다는 의미
    if a % i == 0 and b % i == 0:
        print("최대공약수는 {}입니다.".format(i))
        break
    else:
        continue
```

### Quiz

    list안의 list의 데이터를 꺼내서 출력해보기

```python
member = [['VVIP', 50_000_000], ['VIP', 30_000_000], ['Platinum', 20_000_000], ['Family', 10_000_000]]
member
```

    index를 이용해서 출력해보기
    for, if, elif, else
        VIP이상은 발렛파킹
        Platinum은 우대주차장
        Family 지하 12층

```python
grade = int(input("연간 백화점 소비액을 입력해주세요 : "))
for i in range(1, 4):
    if grade >= member[i][1]:
        print("당신은 발렛파킹 대상자 이십니다.")
        break
    elif grade >= member[i][1]:
        print("당신은 우대주차장 대상자 이십니다.")
        break
    else:
        print("지하 12층으로 가주시길 바랍니다.")
        break
```
