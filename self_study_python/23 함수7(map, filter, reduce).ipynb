## map, filter, reduce

### map

*   순서가 있는 데이터 집합에서 모든 값에 함수를 적용시켜 결과 출력

```python
ls = [1, 2, 3, 4]
def odd_even(n):
    return "odd" if n%2 else "even"
```

```python
# map("함수", "데이터셋")
list(map(odd_even, ls))
```

### quiz : map

    * map을 이용하여 입력받는 숫자를 리스트에 바로 담기
    * input(), list(), split() 사용하기

```python
num = input("input numbers : ")
res = num.split(" ")
res = list(map(int, res))
```

```python
numbers = list(map(int, input("insert numbers : ").split(" ")))
numbers
```

### filter

    * list 데이터에서 특정 조건에 맞는 것만 출력

```python
#filter(함수, 데이터) # Lambda 간단하게 선언
ls = range(10)
ls
```

```python
list(filter(lambda x: True if x%2 else False, ls))
```

## reduce

    * list 데이터를 처음부터 순서대로 특정 함수를 실행하여 결과를 누적

```python
from functools import reduce
ls = [3, 1, 2, 4, 5]
# reduce(함수, 데이터)
reduce(lambda x, y: x + y, ls)
```
