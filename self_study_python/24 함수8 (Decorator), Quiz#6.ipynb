# Decorator

*   함수에서 코드를 바꾸지 않고 기능을 추가하거나 수정하고 싶을 때 사용
*   inner function을 선언했을 때 인스턴트처럼 사용 -> class의 개념
*   큰 틀을 만들어두고 함수를 선언해서 집어넣기

```python
def mother(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result
    return wrapper

def son1():
    return True

def son2():
    return False
```

````python
def plus(a, b):
    print("start!")
    res = a + b
    print(res)
    print("result!")
    return res
````

````python
plus(4, 5)
````

````python
def minus(a, b):
    print("start!")
    res = a - b
    print(res)
    print("result!")
    return res
````

````python
def disp(func):
    def wrapper(*args, **kwargs):
        print("your function getting start!")
        res = func(*args, **kwargs)
        print(res)
        print("process done!")
        return res
    return wrapper
````

````python
@disp
def plus(a, b):
    res = a + b
    return res
````

````python
plus(3, 4)
````

## decorator 실행 예제

*   함수의 실행 시간을 출력하는 함수 작성

````python
def timer(func):
    import time
    def wrapper(*args, **kwargs):
        start_time = time.time()  # 실행된 시간이 들어감
        res = func(*args, **kwargs)
        end_time = time.time()
        print("Running Time: {}".format(end_time - start_time))
        return res
    return wrapper
````

````python
def test1(n1, n2):
    data = range(n1, n2 + 1)
    return sum(data)
````

````python
def test2(n1, n2):
    res = 0
    for n in range(n1, n2 + 1):
        res += n
    return res
````

````python
test1(1, 100_000)
````

````python
test2(1, 100_000)
````

````python
@timer
def test1(n1, n2):
    data = range(n1, n2 + 1)
    return sum(data)
````

````python
@timer
def test2(n1, n2):
    res = 0
    for n in range(n1, n2 + 1):
        res += n
    return res
````

````python
test1(1, 100_000)
````

````python
test2(1, 100_000)
````

## 1.2 예제

*   password가 일치할 때 연산이 가능하도록 decorator 만들기

````python
def check_pw(func):
    def wrapper(*args, **kwargs):
        pw = "euc21"
        login = input("password? ")
        if login == pw : return func(*args, **kwargs)
        return "Wrong password"
    return wrapper
````

````python
@check_pw
def hap(a, b):
    return a + b
````

````python
hap(3, 4)
````

````python
hap(3, 4)
````

# 2 Quiz

*   dict형의 id와 password datas를 생성하고 datas안에 id와 password가 일치하는 경우에만 실행되도록 생성
*   datas는 dict형
*   "id":"password" 형식의 key, value 설정
*   decorator로 함수 실행하도록 생성
*   없거나 틀렸을 경우 접속이 불가능하다는 결과 출력

````python
def datas_conf(func):
    def wrapper(*args, **kwargs):
        datas = {"1234":"1234"}
        login_ID = input("ID : ")
        login_PW = input("PW : ")
        login = {login_ID:login_PW}
        if login == datas:
            return func(*args, **kwargs)
        else:
            return "접속이 불가합니다."
    return wrapper
````

````python
@datas_conf
def hap(a, b):
    return a + b
````

````python
hap(3, 5)
````
