-   월간 주식 수익률 데이터프레임 생성 후 agg() 함수로 분석

```python
import pandas as pd
import numpy as np
dates = pd.date_range(start = "20200101", end = '20201201', freq = 'MS')
dates = pd.Series(dates, name = 'Date')
dates
```

```python
values = np.array([round(np.random.normal(0., 5.,), 2) for _ in range(48)]).reshape(12, -1)
values = pd.DataFrame(values, columns = list('Chocopie Miz Kid-0 Diget'.split(" ")))
values
```

```python
stock_df = pd.concat([dates, values], axis = 1)
stock_df
```

```python
stock_df1 = stock_df[['Chocopie', 'Miz', 'Kid-0', 'Diget']]
stock_df1
```

## 그룹함수 직접 구현하기

```python
stock_df1.describe()
```

```python
pd.DataFrame({'Sum': stock_df1.agg('sum'),
              "Mean": round(stock_df1.agg('mean'), 2),
              'Min': stock_df1.agg('min'),
              'Max': stock_df1.agg('max')}).T
```

## 결측치 채우기

### 결측 데이터 생성

```python
couque_date = pd.to_datetime(['2020-01-01', '2020-02-01', '2020-04-01', '2020-06-01', '2020-07-01', '2020-10-01', '2020-12-01'])
couque_date = pd.Series(couque_date, name = 'Date')
couque_date
```

```python
couque_rev = pd.Series([round(np.random.normal(0., 5.,), 2) for _ in range(len(couque_date))], name = "Couque D'asse")
couque_df = pd.concat([couque_date, couque_rev], axis = 1)
couque_df
```

```python
stock_df
```

```python
stock_df2 = pd.merge(stock_df, couque_df, how = 'outer')
stock_df2
```

```python
stock_df2.index = stock_df2["Date"]
stock_df2.drop("Date", axis = 1, inplace=True)
stock_df2
```

```python
stock_df2.fillna(stock_df2["Couque D'asse"].mean(), inplace=True)
stock_df2
```

```python
pd.options.display.float_format = '{:.2f}'.format
stock_df2
```

```python
stock_df2.describe()
```

```python
pd.DataFrame({'Sum': stock_df2.agg('sum'),
              "Mean": stock_df2.agg('mean'),
              'Min': stock_df2.agg('min'),
              'Max': stock_df2.agg('max')}).T
```
