```python
tit = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/titanic.csv", sep = '\t')
tit
```

## Sex, Pclass 에 따른 데이터 수

```python
tit_df = tit.groupby(["Sex", "Pclass"]).size().reset_index(name = "counts")
tit_df
```

```python
res1 = tit_df.pivot(index = "Sex", columns = "Pclass", values = "counts")
res1
```

```python
tit["Counts"] = 1
tit
```

```python
res2 = tit.pivot_table("Counts", "Sex", "Survived", aggfunc = np.sum)
res2
```

```python
res3 = tit.pivot_table("Counts", ["Pclass"], ["Survived"], aggfunc=np.sum)
res3
```

```python
res3["Total1"] = res3[0] + res3[1]
res3
```

```python
res3.loc["Total"] = res3.loc[1] + res3.loc[2] + res3.loc[3]
res3  # iloc[]와 다름.
```
