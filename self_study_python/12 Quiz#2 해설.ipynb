### Quiz

    list안의 list의 데이터를 꺼내서 출력해보기

    index를 이용해서 출력해보기
    for, if, elif, else
        VIP이상은 발렛파킹
        Platinum은 우대주차장
        Family 지하 12층

```python
member = [['VVIP', 50_000_000], ['VIP', 30_000_000], ['Platinum', 20_000_000], ['Family', 10_000_000]]

for grade in member:
    if grade[1] >= 30_000_000:
        print("{}등급입니다.".format(grade[0]))
        print("발렛파킹이 가능합니다.")
    elif grade[1] >= 20_000_000:
        print("{}등급입니다.".format(grade[0]))
        print("우대주차장 이용이 가능합니다.")
    elif grade[1] >= 10_000_000:
        print("{}등급입니다.".format(grade[0]))
        print("지하 4층까지 이용 가능합니다.")
    else:
        print("지하 12층으로 가세요.")
```
