# DataFrame

*   여러개의 Series로 구성

## 생성

### dict안에 list로 value 생성`

```python
data = {
    "name": ["easyup", "python"],
    "class": ["open", "open2"]
}
data
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(data)
df
```

```python
print(df)
```

### list안에 dict 생성

```python
data2 = [{"name": "easyup", "class": "open"},
         {"name": "python", "class": "open2"}]
data2
```

```python
df2 = pd.DataFrame(data2)
df2
```
