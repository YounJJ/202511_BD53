*   dict형의 id와 password datas를 생성하고 datas안에 id와 password가 일치하는 경우에만 실행되도록 생성
*   datas는 dict형
*   "id":"password" 형식의 key, value 설정
*   decorator로 함수 실행하도록 생성
*   없거나 틀렸을 경우 접속이 불가능하다는 결과 출력

```python
def datas_conf(func):
    def wrapper(*args, **kwargs):
        datas = {"1234":"1234"}
        login_ID = input("ID : ")
        login_PW = input("PW : ")
        login = {login_ID:login_PW}
        if login == datas:
            return func(*args, **kwargs)
        else:
            return "접속이 불가합니다."
    return wrapper
```

```python
@datas_conf
def hap(a, b):
    return a + b
```

```python
# 강사님 답안
def check_pwd(func):
    def wrapper(*args, **kwargs):
        datas = {
            "test":"1234",
            "admin":"euc21",
            "bekko":"easyup"
        }
        user_id, user_pw = input("Enter your id: "), input("Enter password: ")
        for user ,pw in datas.items():
            if (user_id==user) and (user_pw == pw): return func(*args, **kwargs)
        return "Not available to access"
    return wrapper
```

```python
@check_pwd
def hap(a, b):
    return a+b
```

```python
hap(3, 4)
```

```python
hap(2, 5)
```

```python
@check_pwd
def jegophap(a, b):
    return a**2 + b**2
```

```python
jegophap(3, 4)
```
