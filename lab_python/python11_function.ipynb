## Ex. 1

*   함수이름: make_list
*   기능: [start, end) 범위의 정수 난수 n개를 아이템으로 갖는 리스트를 반환.
*   파라미터: start(정수), end(정수), n(0 또는 양의 정수)
*   반환값: 정수 난수들의 리스트.

````python
import random
````

````python
def make_list(start, end, n):
    """
    [start, end) 범위의 정수 난수 n개를 아이템으로 갖는 리스트를 반환.

    @param start - int. 난수 시작 범위 (포함).
    @param end - int. 난수 종료 범위 (미포함).
    @param n - 리스트에 추가할 아이템 개수(난수 개수). 0 또는 양의 정수.
    @random - 정수 난수 n개를 갖는 리스트.
    """
    randlist = [random.randrange(start, end) for _ in range(n)]
    return randlist
````

````python
# 강사님 답안
def make_list(start, end, n):
    """
    [start, end) 범위의 정수 난수 n개를 아이템으로 갖는 리스트를 반환.

    @param start - int. 난수 시작 범위 (포함).
    @param end - int. 난수 종료 범위 (미포함).
    @param n - 리스트에 추가할 아이템 개수(난수 개수). 0 또는 양의 정수.
    @random - 정수 난수 n개를 갖는 리스트.
    """

    # numbers = []
    # for _ in range(n):
    #     numbers.append(random.randrange(start, end))
    numbers = [random.randrange(start, end) for _ in range(n)]

    return numbers
````

````python
rand_numbers = make_list(0, 10, 5)
print(rand_numbers)
````

## Ex. 2

*   함수 이름: calculate_sum
*   기능: 숫자(정수 또는 실수)들을 원소로 갖는 리스트를 전달받아서, 그 리스트의 모든 원소들의 합을 리턴.
*   파라미터: array(숫자들의 리스트).
*   반환값: 리스트 array의 모든 원소들의 합계.

````python
def calculate_sum(array):
    """
    숫자(정수 또는 실수)들을 원소로 갖는 리스트를 전달받아서, 그 리스트의 모든 원소들의 합을 리턴.

    @param array - 숫자들을 원소로 갖는 리스트.
    @return - 리스트 array의 모든 원소들의 합계.
    """
    sum_array = 0
    try:
        for i in array:
            if type(i) in (int, float):
                sum_array += i
    except:
        print("Error!")
    else:
        return sum_array
````

````python
# 강사님 답안

def calculate_sum(array: list):
    """
    숫자(정수 또는 실수)들을 원소로 갖는 리스트를 전달받아서, 그 리스트의 모든 원소들의 합을 리턴.

    @param array - 숫자들을 원소로 갖는 리스트.
    @return - 리스트 array의 모든 원소들의 합계.
    """

    if type(array) != list:
        raise ValueError

    total = 0
    for x in array:
        total += x

    return total
````

*   기능: 숫자들을 원소로 갖는 리스트의 원소들의 평균을 리턴.
*   파라미터: array(숫자들의 리스트).
*   반환값: 리스트 array의 모든 원소들의 평균.

````python
def calculate_mean(array):
    """
    숫자들을 원소로 갖는 리스트의 원소들의 평균을 리턴.

    @param array - 숫자들의 리스트
    @return - 리스트 array의 모든 원소들의 평균
    """
    avg = sum(array) / len(array)
    return avg
````

````python
# 강사님 답안
def calculate_mean(array):
    """
    숫자들의 리스트 arrary의 원소들의 평균을 리턴.

    @param array - 숫자들을 원소로 갖는 리스트.
    @return array - 원소들의 평균.
    """

    mean = calculate_sum(array) / len(array)

    return mean
````

## Ex. 4

*   함수 이름: find_min_and_max
*   기능: 숫자들을 원소로 갖는 리스트에서 최솟값과 최댓값을 찾아서 리턴.
*   파라미터: array(숫자들의 리스트).
*   반환값: 리스트 array에서 최솟값, 최댓값.

````python
def find_min_and_max(array):
    return min(array), max(array)
````

````python
# 강사님 답안
def find_min_and_max(array):
    """
    숫자들의 리스트 array 원소들 중에서 최솟값과 최댓값을 리턴.

    @param array - 숫자들을 저장하는 리스트.
    @return 최솟값, 최댓값
    """

    min_value = array[0]
    max_value = array[0]
    for x in array:
        if x > max_value:
            max_value = x
        if x < min_value:
            min_value = x

    return min_value, max_value
````

# Default Argument(기본 아규먼트)

*   함수를 선언할 때 파라미터에 설정하는 기본값.
    *   default parameter: 기본값이 설정된 파라미터(매개변수).
*   함수를 호출할 때 default parameter에 아규먼트를 주지 않으면, default argument가 기본값으로 사용됨.
*   함수를 호출할 때 default parameter에 아규먼트를 주면, 기본값 대신 전달한 아규먼트가 사용됨.

````python
print('안녕', 'Hello', '파이썬', 'Python')  # sep = ' '기본값을 사용
````

````python
print('안녕', 'Hello', '파이썬', 'Python', sep = ', ')  # sep 파라미터의 기본값을 변경.
````

````python
# default argument를 갖는 함수 정의
def repeat_message(msg, n = 1):
    for _ in range(n):
        print(msg)
````

함수를 호출할 때 아규먼트를 전달하는 방법:

*   **positional argument**: 선언된 파라미터 순서대로 아규먼트들을 전달하는 방식.
*   **keyword argument**: `param = value` 형식으로 아규먼트(들)을 전달하는 방식.
    *   keyword argument 방식은 파라미터 순서를 지키지 않아도 됨.
    *   **(주의)** 함수를 호출할 때 positional argument와 keyword argument를 함께 사용하는 경우에는, 반드시 positional argument들을 먼저 써야 함.
    *   keyword argument를 positional argument보다 먼저 쓸 수는 없음!

# 가변길이 인수(variable-length argument)

*   함수를 호출할 때 아규먼트를 몇 개를 전달해도 됨.
    *   아규먼트를 한 개도 전달하지 않아도 됨.
*   함수를 선언할 때 파라미터 이름 앞에 `*`를 사용한 파라미터가 가변길이 인수가 됨.
*   함수 내부에서 가변길이 인수는 튜플(tuple)로 취급함.
    *   가변길이 인수는 인덱스를 가짐.
    *   가변길이 인수는 `for-in` 구문에서 사용할 수 있음.
*   함수를 선언할 때 가변길이 인수는 1개만 선언할 수 있음.
*   가변길이 인수는 keyword argument 방식으로 아규먼트들을 전달할 수 없음.
*   가변길이 인수 뒤에서 선언된 파라미터(들)은 keyword argument 방식으로만 아규먼트를 전달할 수 있움.

````python
def calculate(*values, op):
    """
    op = '+'인 경우, values 들의 합계를 리턴.
    op = '*'인 경우, values 들의 곱을 리턴.
    (예)
    calculate(1, 2, 3, op = '+') 호출하면 1 + 2 + 3을 리턴.
    calculate(1, 2, 3, op = '*') 호출하면 1 * 2 * 3을 리턴.
    values에 전달된 값이 없을 때는 None을 리턴. (len을 계산해 보면 굳)
    op가 '+' 또는 '*'가 아닐 때는 ValueError를 발생(raise ValueError).
    """
    if len(values) == 0:
        return None
    else:
        if op == "+":
            return sum(values)
        elif op == "*":
            mul_val = 1
            for i in values:
                mul_val *= i
            return mul_val
        else:
            raise ValueError("아규먼트 op는 '+' 또는 '*'만 가능합니다.")
````

````python
calculate(op = '+')
````

````python
calculate(1, 2, 3, 4, 5, op = "+")
````

````python
calculate(1, 2, 3, 4, 5, op = "*")
````

````python
calculate(1, 2, 3, 4, 5, op = '/')
````

````python
calculate(op = '-')
````

# 가변길이 키워드 아규먼트(variable-length keyword arguments)

*   함수를 선언할 때 파라미터 이름 앞에 `**`를 사용.
*   함수를 호출할 때, `param1 = value1, param2 = value2, ...` 형식으로 아규먼트들을 전달하면 됨.
    *   가변길이 --> 전달하는 아규먼트 개수는 제한이 없음.
    *   키워드 아규먼트 --> 반드시 키워드(파라미터 이름)와 함께 아규먼트들을 전달해야 됨.
        *   키워드는 자유롭게 만듦.
*   함수 내부에서 가변길이 키워드 아규먼트는 dict 타입 객체로 취급.
    *   키워드(파라미터 이름)는 dict의 key가 됨.
    *   아규먼트는 dict의 value가 됨.
*   가변길이 키워드 아규먼트는 함수에서 오직 1개만 선언할 수 있음.

````python
def test_kwargs(*args, **kwargs):
    print(f'args = {args}')
    print(f'kwargs = {kwargs}')
````

````python
test_kwargs(1, 2, 'a', 'b')
````

````python
test_kwargs(1, 2, name = '오쌤', age = 16)
````

````python
def test_kwargs2(**kwargs, *arg):
    pass
````

````python
def make_student_info(id, name, **kwargs):
    student = {'id': id, 'name': name}
    print(f'kwargs = {kwargs}')
    for k, v in kwargs.items():
        student[k] = v

    return student
````

````python
make_student_info(1, '홍길동')
````

````python
make_student_info(2, '오쌤', email = 'jake@itwill.com', phone = '010-0000-0000')
````

````python
make_student_info(3, 'Scott', job = 'Manager')
````

````python
def make_student_info2(id, name, **kwargs):
    student = {'id': id, 'name': name}
    for k, v in kwargs.items():
        if k in ('email', 'grade'):
            student[k] = v

    return student
````
