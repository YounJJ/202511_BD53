*   파일의 첫번째 줄은 컬럼 이름들이기 때문에, 첫번째 줄은 읽고 무시.
*   파일의 두번째 줄부터 파일 끝(EOF)까지 반복.
    *   각 줄에서 읽은 내용을 쉼표(,)로 분리한 리스트를 빈 리스트에 append.

        `[['1', '1', '50', '98', '50'], ['2', '1', '60', '97', '60'], ...]`

*   위에서 만들어진 2차원 리스트는 문자열들을 저장하고 있는 리스트이기 때문에, 리스트의 모든 값들을 정수(int)로 변환

    `[[1, 1, 50, 98, 50], [2, 1, 60, 97, 60], ...]`

*   수학 점수들로만 이루어진 1차원 리스트를 만듦.

    `[50, 60, ...]`

*   수학 점수의 최댓값, 최솟값, 평균 출력.
*   1반 학생들의 수학 점수 총점, 평균, 최댓값, 최솟값 출력

```python
file_location = '/content/drive/MyDrive/lab_python/exam.csv'

with open(file = file_location) as f:
    ls1 = [line.strip().split(",") for line in f][1:]

ls_res = [[int(j) for j in i] for i in ls1]
ls_res2 = [x[2] for x in ls_res]

print(f'수학 점수의 [최댓값은: {max(ls_res2)}, 최솟값은: {min(ls_res2)}, 평균은: {(sum(ls_res2) / len(ls_res2))}]')

class1_math = [i[2] for i in ls_res if i[1] == 1]
print(f'1반 학생들의 수학점수 [총점은: {sum(class1_math)}, 평균은: {sum(class1_math) / len(class1_math)}, 최댓값은: {max(class1_math)}, 최솟값은: {min(class1_math)}]')
```

```python
math_byclass = [[i[2] for i in ls_res if i[1] == k] for k in range(1, 6)]

avg_math = [(sum(i)/len(i)) for i in math_byclass]

print(*(f'{i}반 수학 평균 = {avg_math[i-1]}' for i in range(1, 6)), sep='\n')
```

```python
class Student:
    def __init__(self, id, cls, math, english, science):
        self.id = int(id)
        self.cls = int(cls)
        self.math = int(math)
        self.english = int(english)
        self.science = int(science)

    def __repr__(self):
        return f'Student(id = {self.id}, class = {self.cls}, math = {self.math}, english = {self.english}, science = {self.science})'
```

```python
Student('1', '1', '50', '98', '50')
```

```python
# 파일에서 읽은 내용을 Student 객체로 만들어서 리스트에 저장.
exam_student = []

with open(file = file_location) as f:
    f.readline()
    for line in f:
        row = line.strip().split(',')
        student = Student(row[0], row[1], row[2], row[3], row[4])
        exam_student.append(student)

exam_student
```

```python
math = [s.math for s in exam_student]
print(math)
```

```python
math_class1 = [s.math for s in exam_student if s.cls == 1]
math_class1
```

```python
# 반별 과학 점수 평균 출력
for c in range(1, 6):
    science_by_class = [s.science for s in exam_student if s.cls == c]
    mean = sum(science_by_class) / len(science_by_class)
    print(f'{c}반 과학 점수 평균 = {mean}')
```
