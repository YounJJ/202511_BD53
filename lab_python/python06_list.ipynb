# list

*   `list`: 여러 개의 값들을 하나의 변수에 저장하기 위한 데이터 타입.
*   아이템(item). 원소(element): 리스트에 저장된 각각의 값.
*   인덱스(index): 리스트의 아이템들이 저장된 위치를 표시하는 정수.
    *   리스트의 인덱스는 **0**부터 시작.
    *   리스트 마지막 인덱스는 (아이템 개수 - 1)
    *   리스트의 끝에서부터 역순으로 인덱스를 매길 때는 *음수 인덱스*를 사용할 수 있음.
*   리스트는 아이템들을 추가, 변경, 삭제할 수 있음.
    *   `tuple`과 다른 점.

````python
# 숫자들을 저장하는 list
numbers = [1, 2, 10 ,20, -1, -100]
print(type(numbers))
print(numbers)
````

````python
# 파이썬의 리스트는 여러가지 타입의 값들을 하나의 변수에 저장할 수 있음.
values = [1, 2, '안녕', '파이썬', 3.141592, True, [1, 2, 3]]
print(type(values))
print(values)
````

# Slicing

`list_name[start:end]`

*   리스트에서 start <= index < end 범위의 아이템들을 잘라내는 방법.
*   시작 인덱스 start를 생략하면 첫번째 아이템부터 잘라냄. start의 기본값은 0.
*   종료 인덱스 end를 생략하면 마지막 아이템까지 잘리냄.
*   slicing의 결과는 *새로운 리스트*.


````python
numbers
````

````python
numbers[1:4]
````

````python
numbers[3:6]
````

````python
numbers[3:]  # 인덱스 3번 위치부터 끝까지 자름.
````

````python
numbers[:3]  # 리스트에서 첫 3개 원소를 자름.
````

````python
numbers[-2:]  # 리스트에서 마지막 2개 원소를 자름.
````

````python
numbers[:1]  # 리스트에서 첫번째 원소만 잘라낸 새로운 "리스트"
````

## Ex. 1

*   빈 리스트(empty list) numbers를 만듦.
*   리스트 numbers에 0 이상 100 미만의 미만의 정수 난수 10개를 저장(append).
*   리스트 numbers를 출력
*   numbers의 원소들 중에서 짝수들만 골라낸 새로운 리스트 evens를 만들고 출력.
*   numbers의 원소들 중에서 홀수들만 골라낸 새로운 리스트 odds를 만들고 출력.

````python
numbers = []
evens = []
odds = []

import random
while len(numbers) < 10:
    numbers.append(random.randrange(0, 100))
print(numbers)

for i in numbers:
    if not i%2:
        evens.append(i)
    else:
        odds.append(i)

print(evens, odds)
````

````python
# 강사님 답안
import random

numbers = []  # 빈 리스트 만듦.
for _ in range(10):  # 10번 반복
    # r = random.randrange(100)  # random.randrange(0 ,100), 난수 만듦
    # numbers.append(r)  # 난수를 리스트의 끝에 추가
    numbers.append(random.randrange(100))

print(numbers)
````

````python
evens = []  # 짝수들만 저장하기 위한 빈 리스트
for x in numbers:  # 리스트 numbers의 아이템들을 처음부터 끝까지 순서대로 반복하면서
    if x % 2 == 0:  # 그 아이템이 짝수이면
        evens.append(x)  # 리스트 evens에 아이템을 저장.

print(evens)
````

````python
odds = []  # 홀수들만 저장하기 위한 빈 리스트
for x in numbers:  # numbers의 아이템들을 처음부터 끝까지 순서대로 반복하면서
    if x % 2:  # if x % 2 == 1: x가 홀수이면
        odds.append(x)  # odds에 아이템을 추가.

print(odds)
````

## Ex. 2
*   integers 이름의 빈 리스트를 만듦.
*   리스트 integers에 -10 이상 10 이하의 정수 난수 10개를 저장하고 출력.
*   is_positives 이름의 빈 리스트를 만듦.
*   integers의 원소가 양수이면 True, 그렇지 않으면 False를 is_positives에 저장하고 출력.

````python
integers = []

import random
while len(integers) < 10:
    integers.append(random.randrange(-10, 11))
print(integers)

is_positives = []

for i in integers:
    if i >= 0:
        is_positives.append(True)
    else:
        is_positives.append(False)

print(is_positives)
````

````python
# 강사님 답안
integers = []  # 정수 난수들을 저장하기 위한 빈 리스트
for _ in range(10):  # 10번 반복
    integers.append(random.randrange(-10, 11))  # integers에 난수 추가.

print(integers)
````

````python
is_positives = []  # True/False를 저장하기 위한 빈 리스트
for x in integers:  # 리스트 integers의 원소들을 순서대로 반복하면
    if x > 0:
        is_positives.append(True)
    else:
        is_positives.append(False)

print(is_positives)
````

````python
# [1, 10] 사이의 정수들의 제곱을 저장하는 리스트: [1, 4, 9, ..., 81, 100]
# squares = []
# for x in range(1, 11):
#     squares.append(x ** 2)
squares = [x ** 2 for x in range(1, 11)]
squares
````

````python
# 0 이상 100 미만의 정수 난수 10개를 저장하는 리스트
rand_numbers = [random.randrange(100) for _ in range(10)]
rand_numbers
````

````python
# rand_numbers의 원소들 중 짝수들로만 이루어진 리스트
evens = [x for x in rand_numbers if x % 2 == 0]
evens
````

````python
# rand_numbers의 원소들 중 홀수들로만 이루어진 리스트
odds = [x for x in rand_numbers if x % 2]
odds
````

````python
# rand_numbers의 숫자가 짝수이면 True, 홀수이면 False를 저장하는 리스트.
is_even = []
for x in rand_numbers:
    if x % 2 == 0:
        is_even.append(True)
    else:
        is_even.append(False)

is_even
````

# `zip()` 함수와 `for` 반복문

````python
# for i, v in enumerate(list):
````

````python
numbers1 = [1, 2, 3, 4, 5]
numbers2 = [11, 21, 31, 41, 51]

for x, y in zip(numbers1, numbers2):
    print(x, y)
````

````python
# numbers1과 numbers2의 같은 인덱스의 원소들끼리의 덧셈 결과를 저장하는 리스트
# 결과: [12, 23, 34, 45, 56]

result = []
for x, y in zip(numbers1, numbers2):
    result.append(x + y)
result
````

````python
result = [x + y for x, y in zip(numbers1, numbers2)]
result
````
