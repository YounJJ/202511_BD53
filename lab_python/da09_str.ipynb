# 데이터 정제

자동차 회사 이름의 오타 수정:

*   chevroelt --> chevrolet
*   maxda --> mazda
*   mercedes --> mercedes-benz
*   toyouta --> toyota
*   vokswagen --> volkswagen
*   vw --> volkswagen

```python
mpg.loc[mpg.company == "chevroelt", "company"] = "chevrolet"
```

```python
correction_dict = {
    "chevroelt": "chevrolet",
    "maxda": "mazda",
    "mercedes": "mercedes-benz",
    "toyouta": "toyota",
    "vokswagen": "volkswagen",
    "vw": "volkswagen"
}

mpg['company'] = mpg['company'].replace(correction_dict)

mpg['company'].unique().size
```

회사 이름의 빈도수

```python
mpg.company.value_counts()
```

```python
plt.figure(figsize=(16, 10))
sns.countplot(data=mpg, y="company", order=mpg['company'].value_counts().index)

plt.show()
```

회사별 연비 평균 상위/하위 10개 회사

```python
mpg.groupby(by=["company"]).mpg.mean().sort_values().head(10)
```

```python
bottom10 = mpg.groupby(by=["company"]).mpg.mean().sort_values().head(10)

plt.barh(y=bottom10.index, width=bottom10.values)
plt.xlabel("mean of mpg")
plt.show()
```

```python
mpg.groupby(by=["company"]).mpg.mean().sort_values(ascending=False).head(10)
```

```python
top10 = mpg.groupby(by=["company"]).mpg.mean().sort_values(ascending=False).head(10)

plt.barh(y=top10.index, width=top10.values)
plt.show()
```

## 영화 장르 분석

### 장르가 Animation인 영화

```python
# 강사님 답안
animations = movies[movies.genres.str.contains("Animation")]
animations
```

```python
len(animations)  # 데이터프레임의 행의 개수
```

```python
# 뮤지컬 애니메이션
movies.genres.str.contains("musical", case=False)
movies.genres.str.contains("animation", case=False)

musical_animations = movies[(movies.genres.str.contains("musical", case=False)) & (movies.genres.str.contains("animation", case=False))]

musical_animations.head(8)
```

```python
#
len(musical_animations)
```

```python
# not은 ~로 표현 가능
musical_animations = movies[(~movies.genres.str.contains("musical", case=False)) & (~movies.genres.str.contains("animation", case=False))]

musical_animations
```

```python
# 로맨틱 코미디 장르(Romance & Comedy)의 영화들
romantic_comedies = movies[(movies.genres.str.contains("Comedy", case=False)) & (movies.genres.str.contains("Romance", case=False))]

romantic_comedies.head()
```

### 가장 많이 등장하는 영화 장르 상위/하위 10개

```python
# 강사님 답안
genres = []
for x in movies.genres:
    genres.extend(x.split("|"))

print(genres)
```

```python
# Series의 메서드(기능)들을 사용하기 위해서 list를 Series로 변환
genres_sr = pd.Series(genres)
genres_sr
```

```python
genres_sr.unique().size
```

```python
genre_count = genres_sr.value_counts()
genre_count
```

```python
# 가장 많이 등장하는 상위 10개 장르
top10 = genre_count.head(10)
top10
```

```python
# sns.barplot(data=top10)  # 세로 막대 그래프
sns.barplot(y=top10.index, x=top10.values)  # 가로 막대 그래프
plt.show()
```

```python
# 하위 10개 장르
bottom10 = genre_count.tail(10)
bottom10
```

```python
sns.barplot(x=bottom10.values, y=bottom10.index)

plt.show()
```

```python
# 가장 많이 등장하는 장르 이름?
genre_count = movies['genres'].str.split('|').explode().value_counts()
genre_count
```

```python
sns.barplot(x=genre_count.values, y=genre_count.index)

plt.show()
```
