# while 반복문

```
[초기화 문장]
while 조건식:
    조건식을 만족하는 동안에 반복할 코드 블록
    [조건을 변경할 수 있는 문장]
```

*   조건식을 만족하는 동안에 블록 내부를 반복해서 실행하게 됨.
*   조건식을 만족하지 못하면(조건식이 False가 되면) 반복을 멈춤.
*   `while 조건식` 다음에는 반드시 콜론(:)을 사용해야 됨.
*   블록 안에서는 들여쓰기 크기를 일정하게 작성.

````python
# 1 ~ 5까지 순서대로 출력
n = 1  # 변수 초기화 - 변수를 선언하고, 최초로 값을 할당(저장)하는 것.
while n <= 5 :
    print(n)
    n += 1
````

````python
# 5 ~ 1까지 내림차순으로 출력
n = 5
while n >= 1:  # while n > 0: 과 같음.
    print(n)
    n -= 1
````

````python
# 2 4 6 8 10 순서로 출력
n = 2
while n <= 10:
    print(n)
    n += 2
````

````python
n = 2
while n <= 10:
    if n % 2 == 0:
        print(n)
    n += 1  # 만약 이 줄을 한번더 들여쓰면 n = 3 인 상태에서 무한 루프에 빠짐.
````

````python
# 9 7 5 3 1 순서로 출력
n = 9
while n >= 1:
    print(n)
    n -= 2
````

````python
n = 9
while n > 0:
    # if n % 2 == 1: # n을 2로 나눈 나머지가 1과 같으면
    if n % 2:  # n을 2로 나눈 나머지가 있으면
        print(n)
    n -= 1
````

# ` break`와 `continue`

*   `break`: 반복(iteration)을 종료. break가 포함된 가장 가까운 반복문만 빠져나감.
*   `continue`: 다음 반복(iteration)을 계속 수행. continue 아래의 문장들은 건너뛰고, 그 다음 반복을 수행.

````python
n = 1
while n < 6:
    if n == 3:
        break
    print(n)
    n += 1
````

````python
n = 0
while n < 5:
    n += 1
    if n == 3:
        continue
    print(n)
````

# 연습문제

## 문제 1. 구구단 2단 출력

````python
n = 1
while n <= 9:  # while n < 10:
    print('2 x', n, '=', 2 * n)
    n += 1
````

## 문제 2. 구구단 2단부터 9단까지 출력

````python
k = 2

while k <= 9:
    n = 1
    while n <= 9:
        print(k, 'x', n, '=', k * n)
        n += 1
    k += 1
````

````python
# 강사님 답안
x = 2
while x < 10:
    print(f'=== {x}단 ===')
    y = 1
    while y < 10:
        print(f'{x} x {y} = {x * y}')
        y += 1
    print('---------------')
    x += 1
````

## 문제 3. 아래와 같이 출력하는 코드를 작성하세요.

```
*
**
***
****
*****
```

````python
star = '*'

while len(star) <= 5:
    print(star)
    star += '*'
````

## 문제 4. 아래와 같이 출력하는 코드를 작성하세요.

```
*****
****
***
**
*
```

````python
star = '*'
n = 5
while n >= 1:
    print(star * n)
    n -= 1
````

## 문제 5. 구구단 2단은 2x2까지, 3단은 3x3까지, 4단은 4x4까지, ..., 9단은 9x9까지 출력.

````python
n = 2

while n <= 9:
    k = 1
    while k <= n:
        print(f'{n} x {k} = {n * k}')
        k += 1
    n += 1
````

````python
# 강사님 답안
x = 2
while x < 10:
    print(f'=== {x}단 ===')

    y = 1
    while y < 10:
        print(f'{x} x {y} = {x * y}')
        if x == y:
            break
        y += 1

    print('----------')
    x += 1
````
