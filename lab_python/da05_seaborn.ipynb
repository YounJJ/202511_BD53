## 2개 이상의 Axes 사용하기

# Scatter plot(산점도)

두 변수 간의 상관관계를 시각화.

## `scatterplot()`

```python
# tip ~ total_bill 산점도
sns.scatterplot(data=tips, x="total_bill", y="tip",
                s=50, c="IndianRed", marker="v", alpha=0.7)
#> plt.scatter() 함수의 파라미터들을 그대로 사용할 수 있음.
#> s=마커 크기. c=마커 색깔. marker=마커 종류. alpha=불투명도.
plt.grid()
plt.show()
```

## `regplot()`

regression plot: 산점도 + 선형 회귀(linear regression) 직선.

```python
sns.regplot(data=tips, x="total_bill", y="tip", marker="v",
            scatter_kws={"alpha": 0.4, "s": 20},
            line_kws={"color": "IndianRed"})
#> scatter_kws: 산점도의 설정을 변경하기 위한 dictionary.
#> line_kws: 선형 회귀 직선의 설정을 변경하기 위한 dictionary.
plt.grid()
plt.show()
```

## `jointplot()`

산점도 + 히스토그램

```python
sns.jointplot(data=tips, x="total_bill", y="tip")

plt.show()
```

# Category plot(`catplot()`)

카테고리 별로 Axes 객체를 생성하고, 변수의 통계값을 모든 Axes에 같은 종류의 그래프로 시각화.

*   요일(x축) 별 팁(y축)의 평균 막대 그래프
*   성별(Axes 행)과 흡연여부(Axes 열)별로 Axes들을 나눔.

```python
sns.catplot(data=tips, x="day", y="tip",
            row="sex", col="smoker", kind="bar",
            errorbar=None)
#> estimator 파리미터의 기본값은 "mean".
plt.show()
```

```python
tips.groupby(by=["sex", "smoker", "day"], observed=True).tip.mean()
```

## Ex. 1

*   흡연여부(Axes 행), 시간(Axes 열)으로 Axes들을 나눠서, 요일별 팁의 막대 그래프.
*   groupby 메서드와 통계 함수를 사용해서 값 계산.

```python
sns.catplot(data=tips, x="day", y="tip", row="smoker", col="time", kind="bar", errorbar=None)

plt.show()
```

```python
tips.groupby(by=["smoker", "time", "day"], observed=False).tip.mean()
```

## Ex. 2

*   흡연여부(Axes 행), 시간(Axes 열)으로 Axes들을 나눠서, 요일(x축)별 성별(색깔) 팁(y축)의 평균 막대 그래프.
*   groupby를 사용한 값 계산

```python
sns.catplot(data=tips, x="day", y="tip", hue="sex", row="smoker", col="time", estimator="mean", kind="bar", errorbar=None)

plt.show()
```

```python
tips.groupby(by=["smoker", "time", "day", "sex"], observed=False).tip.mean()
```

## Ex. 3

*   흡연여부(Axes 행)와 시간(Axes 열)으로 Axes를 나눠서, 요일별 팁의 box plot.
*   groupby, describe 메서드를 사용해서 기술 통계량을 출력.

```python
sns.catplot(data=tips, x="day", y="tip", row="smoker", col="time", kind="box")

plt.show()
```

```python
tips.groupby(by=["smoker", "time", "day"], observed=False).tip.describe()
```

```python
g = sns.catplot(data=tips, x="day", y="tip", kind="box", row="smoker", col="time")

print(g.axes)  #> Axes들을 갖는 2차원 배열.
for row in g.axes:
    for a in row:
        a.grid()

# 모든 Axes의 레이블 표시.
g.tick_params(axis="x", labelbottom=True)
g.tick_params(axis="y", labelleft=True)

plt.show()
```

# Interactive Graph

*   plotly 라이브러리

```python
import plotly.express as px
```

## Box plot

```python
figure = px.box(data_frame=tips, x="sex", y="tip", color="sex")

figure.show()
```

```python
# 요일별 흡연여부별 팁의 box plot
fig = px.box(data_frame=tips, x="day", y="tip", color="smoker")

fig.show()
```

## Scatter plot

```python
# tip ~ total_bill
fig = px.scatter(data_frame=tips, x="total_bill", y="tip")

fig.show()
```

```python
# tip ~ total_bill 산점도.
# 성별로 색깔을 다르게, 요일별로 마커 모양을 다르게, 인원수별로 마커 크기를 다르게 표현

fig = px.scatter(data_frame=tips, x="total_bill", y="tip", color="sex", symbol="day", size="size")

fig.show()
```

## Histogram

```python
figure = px.histogram(data_frame=tips, x="total_bill", nbins=10, template="simple_white")

figure.show()
```

## Facetting

Figure 안에 행과 열을 구분해서 Axes들을 여러개 생성하고 그래프를 시각화.

```python
# 흡연여부와 시간으로 Axes를 나눠서 tip ~ total_bill 산점도.
figure = px.scatter(data_frame=tips, x="total_bill", y="tip", facet_row="smoker", facet_col="time", color="sex")

figure.show()
```

```python
sns.relplot(data=tips, x="total_bill", y="tip", hue="sex", row="smoker", col="time")

plt.show()
```

```python
figure = px.box(data_frame=tips, x="day", y="tip",
                facet_row="smoker", facet_col="time")

figure.show()
```
